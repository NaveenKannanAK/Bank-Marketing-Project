### 6.14 – Statistical Hypothesis Testing

#### A. Chi-Square Test for Categorical Features

```python
from scipy.stats import chi2_contingency

categorical_features = bank_marketing_clean.select_dtypes(
    include="object"
).columns.drop("y", errors="ignore")

chi_square_results = []

for col in categorical_features:

    contingency_table = pd.crosstab(
        bank_marketing_clean[col],
        bank_marketing_clean["y"]
    )

    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )

    chi_square_results.append({
        "Feature": col,
        "Chi-Square": chi2,
        "P-Value": p_value,
        "Degrees of Freedom": dof,
        "Significant": "Yes" if p_value < 0.05 else "No"
    })

chi_square_report = pd.DataFrame(chi_square_results)

chi_square_report = chi_square_report.sort_values(
    by="P-Value"
)

display(chi_square_report.round(5))
```

#### B – Numerical Feature Hypothesis Testing

```python
# 6.14 (B) - Mann-Whitney U Test

from scipy.stats import mannwhitneyu

numerical_test_results = []

for col in numerical_features:

    group_no = bank_marketing_clean[
        bank_marketing_clean["y"] == "no"
    ][col]

    group_yes = bank_marketing_clean[
        bank_marketing_clean["y"] == "yes"
    ][col]

    statistic, p_value = mannwhitneyu(
        group_no,
        group_yes,
        alternative="two-sided"
    )

    numerical_test_results.append({
        "Feature": col,
        "U-Statistic": statistic,
        "P-Value": p_value,
        "Significant": "Yes" if p_value < 0.05 else "No"
    })

numerical_test_report = pd.DataFrame(
    numerical_test_results
)

numerical_test_report = numerical_test_report.sort_values(
    by="P-Value"
)

display(numerical_test_report.round(6))
```

### 6.15 – Categorical Feature Association Analysis (Cramér's V)

```python
# 6.15 - Cramer's V Association Analysis

import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

cramers_results = []

for col in categorical_features:

    contingency_table = pd.crosstab(
        bank_marketing_clean[col],
        bank_marketing_clean["y"]
    )

    chi2, p_value, dof, expected = chi2_contingency(
        contingency_table
    )

    n = contingency_table.sum().sum()
    rows, cols = contingency_table.shape

    cramers_v = np.sqrt(
        chi2 / (n * min(rows - 1, cols - 1))
    )

    cramers_results.append({
        "Feature": col,
        "Chi-Square": chi2,
        "P-Value": p_value,
        "Cramers V": cramers_v
    })

cramers_report = pd.DataFrame(cramers_results)

cramers_report = cramers_report.sort_values(
    by="Cramers V",
    ascending=False
)

display(cramers_report.round(4))
```

```python
# 6.18 - Categorical Subscription Rate Analysis

import pandas as pd

categorical_analysis = []

for col in categorical_features:

    summary = (
        bank_marketing_clean
        .groupby(col, observed=True)["y"]
        .agg(
            Total_Customers="count",
            Subscribed=lambda x: (x == "yes").sum()
        )
        .reset_index()
    )

    summary["Not_Subscribed"] = (
        summary["Total_Customers"] - summary["Subscribed"]
    )

    summary["Subscription_Rate (%)"] = (
        summary["Subscribed"] /
        summary["Total_Customers"] * 100
    )

    summary["Feature"] = col

    summary.rename(columns={col: "Category"}, inplace=True)

    categorical_analysis.append(summary)

# Combine all categorical reports
categorical_report = pd.concat(
    categorical_analysis,
    ignore_index=True
)

# Arrange columns
categorical_report = categorical_report[
    [
        "Feature",
        "Category",
        "Total_Customers",
        "Subscribed",
        "Not_Subscribed",
        "Subscription_Rate (%)"
    ]
]

# Sort by subscription rate
categorical_report = categorical_report.sort_values(
    by="Subscription_Rate (%)",
    ascending=False
)

display(categorical_report.round(2))
```

### 1. Handling Class Imbalance Using Random Oversampling (ROS)

```python
# ============================================
# STEP 1: HANDLE CLASS IMBALANCE
# ============================================

from imblearn.over_sampling import RandomOverSampler
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Check original class distribution
print("BEFORE OVERSAMPLING")
print("-" * 45)

print(y_train.value_counts())

# Step 2: Apply Random Oversampling
ros = RandomOverSampler(
    sampling_strategy='auto',
    random_state=42
)

X_train_resampled, y_train_resampled = ros.fit_resample(
    X_train_scaled,
    y_train
)

# Step 3: Check balanced class distribution
print("\nAFTER OVERSAMPLING")
print("-" * 45)

print(y_train_resampled.value_counts())

# Step 4: Display dataset shapes
print("\nDATASET SHAPES")
print("-" * 45)

print("Original Training Shape:", X_train_scaled.shape)
print("Balanced Training Shape:", X_train_resampled.shape)

print("Testing Shape:", X_test_scaled.shape)

# Step 5: Visualize class distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.countplot(x=y_train, ax=axes[0])
axes[0].set_title("Before Oversampling")
axes[0].set_xlabel("Subscription")
axes[0].set_ylabel("Customer Count")

sns.countplot(x=y_train_resampled, ax=axes[1])
axes[1].set_title("After Oversampling")
axes[1].set_xlabel("Subscription")
axes[1].set_ylabel("Customer Count")

plt.tight_layout()
plt.show()
```

### 2: Machine Learning Model Building

```python
# ==========================================================
# STEP 2: MACHINE LEARNING MODEL BUILDING AND COMPARISON
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import time

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# ----------------------------------------------------------
# 1. Initialize Models
# ----------------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5,
        n_jobs=-1
    ),

    "SVM": LinearSVC(
        max_iter=5000,
        random_state=42
    )
}

# Add XGBoost if available
try:
    from xgboost import XGBClassifier

    models["XGBoost"] = XGBClassifier(
        n_estimators=200,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

except ImportError:
    print("XGBoost is not installed. Skipping XGBoost.")

# ----------------------------------------------------------
# 2. Train Models and Evaluate Performance
# ----------------------------------------------------------

results = []
trained_models = {}
predictions = {}

for name, model in models.items():

    print("\nTraining:", name)

    start_time = time.time()

    # Train model using balanced training data
    model.fit(X_train_resampled, y_train_resampled)

    # Predict testing data
    y_pred = model.predict(X_test_scaled)

    # Get prediction scores for ROC-AUC
    if hasattr(model, "predict_proba"):
        y_score = model.predict_proba(X_test_scaled)[:, 1]

    else:
        y_score = model.decision_function(X_test_scaled)

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    roc_auc = roc_auc_score(y_test, y_score)

    pr_auc = average_precision_score(y_test, y_score)

    training_time = time.time() - start_time

    # Store results
    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "Training Time (sec)": training_time
    })

    trained_models[name] = model
    predictions[name] = y_pred

    print("Completed:", name)

# ----------------------------------------------------------
# 3. Create Model Comparison Report
# ----------------------------------------------------------

model_comparison = pd.DataFrame(results)

model_comparison = model_comparison.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 80)
print("MACHINE LEARNING MODEL COMPARISON")
print("=" * 80)

display(model_comparison.round(4))

# ----------------------------------------------------------
# 4. Visualize Model Performance
# ----------------------------------------------------------

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC"
]

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

for ax, metric in zip(axes.flatten(), metrics):

    sns.barplot(
        data=model_comparison,
        x=metric,
        y="Model",
        ax=ax
    )

    ax.set_title(f"Model Comparison - {metric}")
    ax.set_xlim(0, 1)
    ax.set_xlabel(metric)
    ax.set_ylabel("")

axes.flatten()[-1].axis("off")

plt.tight_layout()
plt.show()

# ----------------------------------------------------------
# 5. Classification Report of Each Model
# ----------------------------------------------------------

for name, y_pred in predictions.items():

    print("\n" + "=" * 70)
    print("CLASSIFICATION REPORT:", name)
    print("=" * 70)

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=[
                "No Subscription",
                "Yes Subscription"
            ],
            zero_division=0
        )
    )

# ----------------------------------------------------------
# 6. Confusion Matrices
# ----------------------------------------------------------

fig, axes = plt.subplots(
    2,
    3,
    figsize=(16, 10)
)

axes = axes.flatten()

for ax, (name, y_pred) in zip(
    axes,
    predictions.items()
):

    cm = confusion_matrix(y_test, y_pred)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        ax=ax
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

for ax in axes[len(predictions):]:
    ax.axis("off")

plt.tight_layout()
plt.show()

print("\nMODEL TRAINING AND EVALUATION COMPLETED!")
```

# KNN Hyperparameter Tuning

```python

# ==========================================================
# KNN HYPERPARAMETER TUNING - CONFIGURATION COMPARISON
# ==========================================================

from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)
import pandas as pd
import time

# ----------------------------------------------------------
# 1. Define KNN Configurations
# ----------------------------------------------------------

knn_configs = {
    "Config 1": 3,
    "Config 2": 5,
    "Config 3": 7,
    "Config 4": 9,
    "Config 5": 11
}

# ----------------------------------------------------------
# 2. Train and Evaluate Each Configuration
# ----------------------------------------------------------

knn_results = []
knn_models = {}
knn_predictions = {}

for config_name, k in knn_configs.items():

    print(f"\nTraining {config_name} (K={k})...")

    # Initialize model
    knn_model = KNeighborsClassifier(
        n_neighbors=k,
        n_jobs=-1
    )

    # Training
    start_time = time.time()

    knn_model.fit(
        X_train_resampled,
        y_train_resampled
    )

    training_time = time.time() - start_time

    # Prediction and prediction timing
    prediction_start = time.time()

    y_pred = knn_model.predict(X_test_scaled)

    prediction_time = time.time() - prediction_start

    # Prediction probabilities for ranking metrics
    y_score = knn_model.predict_proba(X_test_scaled)[:, 1]

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    roc_auc = roc_auc_score(y_test, y_score)

    pr_auc = average_precision_score(y_test, y_score)

    # Store results
    knn_results.append({
        "Configuration": config_name,
        "K Neighbours": k,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "Training Time (sec)": training_time,
        "Prediction Time (sec)": prediction_time
    })

    knn_models[config_name] = knn_model
    knn_predictions[config_name] = y_pred

    print(f"{config_name} completed.")

# ----------------------------------------------------------
# 3. Create Configuration Comparison Report
# ----------------------------------------------------------

knn_results_df = pd.DataFrame(knn_results)

knn_results_df = knn_results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 85)
print("KNN CONFIGURATION COMPARISON")
print("=" * 85)

display(knn_results_df.round(4))

# ----------------------------------------------------------
# 4. Identify Best Configuration
# ----------------------------------------------------------

best_knn_config = knn_results_df.iloc[0]["Configuration"]

best_knn_model = knn_models[best_knn_config]

best_knn_prediction = knn_predictions[best_knn_config]

print("\nBEST KNN CONFIGURATION")
print("=" * 50)
print("Configuration:", best_knn_config)
print("K Neighbours:", int(knn_results_df.iloc[0]["K Neighbours"]))
print("Best F1 Score:", round(knn_results_df.iloc[0]["F1 Score"], 4))

print("\nKNN CONFIGURATION TUNING COMPLETED!")
```

# RANDOM FOREST - THRESHOLD OPTIMIZATION

```python

# ==========================================================
# RANDOM FOREST - PROBABILITY THRESHOLD ANALYSIS
# ==========================================================

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Get subscription probabilities
y_prob_rf = best_rf_model.predict_proba(X_test_scaled)[:, 1]

# Define thresholds
thresholds = [0.20, 0.30, 0.40, 0.50, 0.60, 0.70]

threshold_results = []

for threshold in thresholds:

    # Convert probabilities into predictions
    y_pred_threshold = (y_prob_rf >= threshold).astype(int)

    threshold_results.append({
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_test, y_pred_threshold),
        "Precision": precision_score(
            y_test, y_pred_threshold, zero_division=0
        ),
        "Recall": recall_score(
            y_test, y_pred_threshold, zero_division=0
        ),
        "F1 Score": f1_score(
            y_test, y_pred_threshold, zero_division=0
        )
    })

# Create comparison table
threshold_df = pd.DataFrame(threshold_results)

print("RANDOM FOREST - THRESHOLD COMPARISON")
print("=" * 65)

display(threshold_df.round(4))

# Visualize threshold performance
threshold_plot = threshold_df.melt(
    id_vars="Threshold",
    value_vars=["Precision", "Recall", "F1 Score"],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(10, 6))

for metric in ["Precision", "Recall", "F1 Score"]:

    data = threshold_plot[
        threshold_plot["Metric"] == metric
    ]

    plt.plot(
        data["Threshold"],
        data["Score"],
        marker="o",
        label=metric
    )

plt.title("Random Forest - Threshold Optimization")
plt.xlabel("Classification Threshold")
plt.ylabel("Performance Score")
plt.xticks(thresholds)
plt.ylim(0, 1)
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
